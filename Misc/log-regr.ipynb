```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score
```

```python
data = load_breast_cancer()
df = pd.DataFrame(data=data.data, columns=data.feature_names)
df['target'] = data.target
df.head()
```

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1:]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10)
```

```python
log_reg = LogisticRegression(max_iter=10000)
log_reg.fit(X_train, np.ravel(y_train))
```

```python
preds = log_reg.predict(X_test)
preds[:5]
```

```python
y_test[:5]
```

```python
r2 = r2_score(y_test, preds)
r2
```

```python
### using cross validation

params = [
    {
        'C' : [1,5,10],
        'max_iter' : [10000, 12000]
    }
]

model = LogisticRegression()
cv_model = GridSearchCV(model, param_grid=params, scoring='f1',cv=5)
cv_model.fit(X_train, np.ravel(y_train))
```

```python
cv_model.best_params_
```

```python
cv_model.best_score_
```

```python
from sklearn.metrics import confusion_matrix, classification_report

preds_cv = cv_model.predict(X_test)
conf = confusion_matrix(y_test, preds_cv)

sns.heatmap(conf, annot=True, fmt='d')
plt.xlabel('Predictions')
plt.ylabel('Truth')
```

```python
print(classification_report(y_test, preds_cv))
```
