```python
# input text
text = "I am Batman"
text
```

```python
# splittiing
tokens = text.split()
tokens
```

```python
# vocab

vocab = sorted(set(tokens))
vocab
```

```python
# str to int mapping

stoi = {word :  i for i, word in enumerate(vocab)}
stoi
```

```python
# itos -> reverse mapping for decoding

itos = {i : word for word, i in stoi.items()}
itos
```

```python
# list of encoded tokens -> ints

encoded = [stoi[token] for token in tokens]
encoded
# output: [1,   2,      0]
#         "I", "am", "Batman"
```

```python
# decoded -> i to s

decoded = [itos[i] for i in encoded]
decoded
```
