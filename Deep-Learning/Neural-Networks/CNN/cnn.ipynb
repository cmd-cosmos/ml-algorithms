# CNN
- convolutional neural networks
- deal with images
- better than ANNs as they reduce computation, deal with spread apart pixels as being the same as local pixels by mathching locality
- not sensitive to location of an object in an image
- reson why a CNN is better than an ANN at image recognition is that it does not rely on the position of the obj in the image, it breaks down a test sample to learn the features associated with the object.
- run a filter through the image to extract different features associated with the image
- Eg) koala detection



- Eg) Digit



# Filter:
- used to extract feature, sort of like a sliding window/ kernel capturing different aspects of the image by convolving through the image



- apply a convolution/filter operation on the original image.
- use the different filters found for different features of the image



- filter size is a hyperparameter - can be 2x2, 3x3 , 4x4
- stride is the amount of movement or the slide in a direction ---> usually 1

- if we areclassifying digit images then we can say 9 has a loop on the top and it will have an associated filter, which when convolved through the image should activate to the top of the image and not the bottom or somewhere else.









- reconvoluting using a preexisting feature maps for 2D features




- the first portion of a CNN ---> feature extraction
- the dense layer of the CNN ---> classification operation
- relu used to bring non linearity and replaces any -ve vals to 0
