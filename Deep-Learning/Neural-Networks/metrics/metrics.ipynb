## precision, recall, f1 score, true positives

- true positive ---> correct predictions out of all predictions made
- false positives ---> wrong samples marked with the required category.


- accuracy ---> number of predictions that the model got right, total number of true positives and true negatives


- precision ---> how many true positives out of total true positive classification


- recall ---> how many truths did we get right vs how many many truths actually exist in the samples


- precision and recall are for individual classes.

- f1 score ---> harmonic mean of the precision and recall ---> gives overall health of the model

 

```python
import seaborn as sns
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
```

```python
truth =      ["Dog","Not a dog","Dog","Dog", "Dog", "Not a dog", "Not a dog", "Dog",       "Dog", "Not a dog"]
prediction = ["Dog","Dog", "Dog","Not a dog","Dog", "Not a dog", "Dog", "Not a dog", "Dog", "Dog"]
```

```python
cm = confusion_matrix(truth, prediction)
cm
```

```python
class_names = ["Dog", "Not a Dog"]
df = pd.DataFrame(cm, index=class_names, columns=class_names)
fig = sns.heatmap(df, annot=True)
plt.xlabel("prediction")
plt.ylabel("truth")
fig.xaxis.set_ticklabels(fig.xaxis.get_ticklabels())
fig.yaxis.set_ticklabels(fig.yaxis.get_ticklabels())
fig
```
