```python
def get_unique(df):
    for col in df:
        if df[col].dtypes=='object':
            print("{} : {}".format(col, df[col].unique()))
```

```python
get_unique(df1)
```

```python
### we can replace the verbose No's with a simple No

df1.replace('No phone service','No',inplace=True)
df1.replace('No internet service','No',inplace=True)
```
